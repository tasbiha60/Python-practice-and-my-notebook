```python
class Phone:
  def __init__(self,model,battery=100):
    self.model=model
    self.battery=battery
class SmartPhone(Phone):
  def __init__(self,model,operating_system,battery=100,):
    super().__init__(model,battery)
    self.operating_system=operating_system


class  GamingPhone(SmartPhone):
  def __init__(self,model,operating_system,cooling_system,battery=100):
    super().__init__(model,battery,operating_system)
    self.cooling_system=cooling_system
  def start_game(self,name):
    print(f"Playing {name} on {self.model}")
p=Phone("IPHONE")
s=SmartPhone("MOTORWALA","ANDROID")
g=GamingPhone("BLUEBERRY","SMART","HEAT")
## p.start_game("PS5") parent class can never access child class method
g.start_game("PS5")
print(p.model)
print(g.cooling_system)


### fisrt practice problem of oop of pracice doc on oop python

```

```python
class MobilePhone:
  def __init__(self,model,battery=100):
    self.model=model
    self.__battery=battery
  def battery_getter(self):
    return self.__battery
  def use(self):
    self.__battery-=10
    print(f"battery reduced by {self.__battery}")
  def charge(self):
    self.__battery = min(100, self.__battery + 10) ## 100 er beshi hote parbe na but 100 ba er kom hote parbe
    print(f"Battery increased by {self.__battery}")

m=MobilePhone("Iphone")
print(m.battery_getter())
m.use()
m.charge()
## print(m.battery) private atts directly  access kora jay na

```
