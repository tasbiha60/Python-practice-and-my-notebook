## **Cursor**

```python
with open("./sample_data/sample.txt","r") as file:
 print(file.tell()) ## CURRENTLY CURSOR 0 NUM INDEX E
 print(file.read())
 print(file.tell())
 print(file.read()) ## kuno output asbe na karon 11 theke ar inputnai
```

```python
with open("./sample_data/sample.txt","r") as file:
  print(file.tell()) ## 0 te asi
  print(file.read(5)) ## 5 ghor shamne gelam
  print(file.tell())
  print(file.read()) ## 5 num index theke porbe
  file.seek(0) ## first index e niye jay abar
  print(file.tell())
  print(file.read(5))
```
